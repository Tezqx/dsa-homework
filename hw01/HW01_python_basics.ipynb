````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
cities_clean = []
amounts_int =[]
total = 0
count_of_unpaid = 0
sum_of_unpaid =0
#1
for i in range(len(cities)):
    cities_clean.append(cities[i].strip().capitalize())
for i in range(len(amounts)):
    amounts_int.append(int(amounts[i]))
#2
general_amount_of_orders = len(amounts)
for i in range(len(amounts)):
    total += amounts_int[i]

mean = round(total / general_amount_of_orders)
#4
for i in range (len(paid)):
    if paid[i] == False:
        count_of_unpaid += 1
for amount, is_paid in zip(amounts_int, paid):
    if is_paid == False:
        sum_of_unpaid += amount


print("#1 Список городов в единном виде:", cities_clean)
print("#2 Список цен на заказы в целочисленном виде:", amounts_int)  
print("#3 Количество заказов всего:", general_amount_of_orders)
print("#3 Сумма всех заказаов:", total)
print("#3 Средний чек:", mean) 
print("#4 Количество неоплаченных заказов:", count_of_unpaid)  
print("#4 Сумма неоплаченных заказов:", sum_of_unpaid)  
#5
for city_1, amount_1, status in zip(cities_clean, amounts_int, paid):
    print(f"{city_1} | {amount_1} | {status}") 

   

  
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
import math
astana = 0
almaty = 0
max_value = amounts_int[0]
max_index = None
max_city = None
weekend_amount = 0

#6
for city_1, amount_1, status in zip(cities_clean, amounts_int, paid):
    if city_1 == "Astana" and status:
        astana += amount_1
    if city_1 == "Almaty" and status:
        almaty += amount_1

#7
for i, (city_1, amount_1, status) in enumerate(zip(cities_clean, amounts_int, paid)):
    if amount_1 > max_value and status == True:
        max_value = amount_1
        max_index = i
        max_city = city_1
#8
for days_1, amount_1 in zip(days, amounts_int):
    if days_1 == "сб" or days_1 == "вс":
        weekend_amount += 1 
weekend_amount_pr = round((weekend_amount / len(days)) * 100) 
#9
saved = 0
monthly_deposit = 15000
course_price = 400000
count = 0
count_1 = 0
saved_1 = 0
while saved < course_price:
    saved += 15000
    count += 1

while saved_1 < course_price:
    saved_1 += monthly_deposit
    monthly_deposit += 1000
    
    count_1 += 1
print("#6 Выручка Алмата | ", almaty)  
print("#6 Выручка Астана | ", astana)   
print(f"#7 Индекс и город самого крупного оплаченного заказа: {max_index} | {max_city} | {max_value}т")  
print(f"#8 Доля заказов выходного дня: {weekend_amount_pr}%")        
print("#9 Число месяцев:", math.floor(count))
print("#9 Число месяцев но 1000 добавляется каждый месяц:", count_1)
````
